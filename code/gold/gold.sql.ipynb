## How much Revenue did we generate everyday?

```python
%sql
CREATE OR REPLACE TABLE ecommerce.`03_gold`.gold_daily_revenue AS

    SELECT
        created_time,
        COUNT(DISTINCT order_id) AS total_orders,
        SUM(quantity) AS total_quantity,
        ROUND(SUM(order_amount),2) AS total_revenue
    FROM ecommerce.`02_silver`.silver_orders
    GROUP BY created_time
    ORDER BY created_time
```

#### View Gold Table of Daily Sales

```python
%sql
select * from ecommerce.`03_gold`.gold_daily_revenue
```

## Which Product generate the highest Revenue?

```python
%sql
CREATE OR REPLACE TABLE ecommerce.`03_gold`.gold_top_product AS

    SELECT
        product_name,
        COUNT(DISTINCT order_id) AS total_orders,
        SUM(quantity) AS total_quantity,
        ROUND(SUM(order_amount),2) AS total_revenue
    FROM ecommerce.`02_silver`.silver_orders
    GROUP BY product_name 
    ORDER BY total_revenue DESC
```

#### View Gold Table of Top Product

```python
%sql
select * from ecommerce.`03_gold`.gold_top_product
```

## Which of the best selling SKU?

```python
%sql
CREATE OR REPLACE TABLE ecommerce.`03_gold`.gold_top_sku AS
    SELECT
        seller_sku,
        COUNT(DISTINCT order_id) AS total_orders,
        SUM(quantity) AS total_quantity,
        ROUND(SUM(order_amount),2) AS total_revenue
    FROM ecommerce.`02_silver`.silver_orders
    GROUP BY seller_sku 
    ORDER BY total_revenue DESC
```

#### View Gold Table of Top SKU

```python
%sql
select * from ecommerce.`03_gold`.gold_top_sku
```
