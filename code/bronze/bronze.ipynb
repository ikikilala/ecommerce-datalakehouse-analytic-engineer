# Read CSV File

```python
df = (spark.read
      .option("header", "true")
      .option("inferSchema", "true")
      .csv("/Volumes/ecommerce/01_bronze/data_source/All order-2026-09-15-09_56.csv")
      )
```

```python
df.display()
```

# Rename/Remove invalid Column Names

```python
import re

df = df.toDF(*[
    re.sub(r'[^a-zA-Z0-9_]', '_', col).lower()
    for col in df.columns
])
```

## Sanity Checks on Column Names

```python
df.columns
```

# Write into Bronze Table

```python
df.write \
    .format("delta")\
    .mode("overwrite")\
    .saveAsTable("ecommerce.`01_bronze`.orders")
```
